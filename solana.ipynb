```python
def clean_crypto_data(df, date_col='Дата', close_col='Закрыть'):
    df_clean = df.copy()
    
   
    df_clean[date_col] = pd.to_datetime(df_clean[date_col])
    df_clean.set_index(date_col, inplace=True)
    
   
    object_cols = df_clean.select_dtypes(include=['object']).columns
    
   
    for col in object_cols:
        df_clean[col] = (df_clean[col]
                        .astype(str)
                        .str.replace('$', '')
                        .str.replace(',', '')
                        .str.replace('USD', '')
                        .str.strip())
        
       
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
    
    
    df_clean.sort_index(inplace=True)
    
    return df_clean


ethereum_clean = clean_crypto_data(ethereum)
solana_clean = clean_crypto_data(solana)
tether_clean = clean_crypto_data(tether)


print("Ethereum")
print(ethereum_clean.head())
print(ethereum_clean.info())

print("Solana")
print(solana_clean.head())
print(solana_clean.info())

print("Tether")
print(tether_clean.head())
print(tether_clean.info())
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

n_forecasts = 30  
forecasts_sarimax = {}

for col in ['Ethereum', 'Solana', 'Tether']:
    series = merge[col].dropna()
    print(f"{col}: {len(series)} записей, последняя цена: ${series.iloc[-1]:.2f}")

    try:
        model = SARIMAX(series, 
                        order=(1, 1, 1),      
                        seasonal_order=(1, 1, 1, 7))  
        fitted = model.fit(disp=False)
        pred = fitted.forecast(steps=n_forecasts)
        forecasts_sarimax[col] = pred.values
        
        print(f"  Прогноз через 30 дней: ${pred.iloc[-1]:.2f}")
        print(f"  Изменение: {((pred.iloc[-1] - series.iloc[-1]) / series.iloc[-1] * 100):.2f}%")
    except Exception as e:
        print(f"  ✗ Ошибка: {e}")
        forecasts_sarimax[col] = np.full(n_forecasts, np.nan)
    print()
```

```python
print("SARIMAX прогнозы на 30 дней")
for col, preds in forecasts_sarimax.items():
    last_price = merge[col].iloc[-1]
    print(f"{col}:")
    print(f"Текущая цена: ${last_price:.2f}")
    print(f"Прогноз через 30 дней: ${preds[-1]:.2f}")
    print(f"Ожидаемое изменение: {((preds[-1] - last_price) / last_price * 100):.2f}%")
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

seq_len = 60      
horizon = 10      

scaler = MinMaxScaler()
values = merge[['Ethereum', 'Solana', 'Tether']].dropna().values
values_scaled = scaler.fit_transform(values)  

X, y = [], []

for i in range(seq_len, len(values_scaled) - horizon + 1):
    X.append(values_scaled[i - seq_len : i])       
    y.append(values_scaled[i : i + horizon])       

X = np.array(X)  
y = np.array(y)  
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

values = merge[['Ethereum', 'Solana', 'Tether']].dropna().values

scaler = MinMaxScaler()
values_scaled = scaler.fit_transform(values)

seq_len = 60     
horizon = 10      


X, y = [], []

for i in range(seq_len, len(values_scaled) - horizon + 1):
    X.append(values_scaled[i - seq_len:i])      
    y.append(values_scaled[i:i + horizon])      

X = np.array(X)
y = np.array(y)

print(f"X shape: {X.shape}")  
print(f"y shape: {y.shape}")  

train_size = int(len(X) * 0.8)
X_train = torch.FloatTensor(X[:train_size])
y_train = torch.FloatTensor(y[:train_size])
X_test = torch.FloatTensor(X[train_size:])
y_test = torch.FloatTensor(y[train_size:])

print(f"Train: {X_train.shape}, Test: {X_test.shape}")
```

```python
class SimpleLSTM(nn.Module):
    def __init__(self, input_size=3, hidden_size=64, num_layers=2, horizon=10):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=0.2
        )
        self.fc = nn.Linear(hidden_size, horizon * input_size)
        self.horizon = horizon
        self.input_size = input_size

    def forward(self, x):
        lstm_out, (h, c) = self.lstm(x)
        last_hidden = h[-1]  # берем последний скрытой слой
        out = self.fc(last_hidden)
        out = out.view(-1, self.horizon, self.input_size)
        return out

model = SimpleLSTM(input_size=3, hidden_size=64, num_layers=2, horizon=10)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)


print(f"Параметров: {sum(p.numel() for p in model.parameters()):,}")
```

```python
num_epochs = 50
train_losses = []
test_losses = []

for epoch in range(num_epochs):
    model.train()
    optimizer.zero_grad()
    output = model(X_train)
    loss = criterion(output, y_train)
    loss.backward()
    optimizer.step()
    train_losses.append(loss.item())
    model.eval()
    with torch.no_grad():
        test_output = model(X_test)
        test_loss = criterion(test_output, y_test)
        test_losses.append(test_loss.item())
    
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1:2d}/{num_epochs} | Train Loss: {loss.item():.6f} | Test Loss: {test_loss.item():.6f}")
```

```python
plt.figure(figsize=(12, 5))
plt.plot(train_losses, label='Train Loss', alpha=0.8)
plt.plot(test_losses, label='Test Loss', alpha=0.8)
plt.title('LSTM Training History')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
model.eval()
with torch.no_grad():
    train_pred = model(X_train).numpy()
    test_pred = model(X_test).numpy()

train_actual = y_train.numpy()
test_actual = y_test.numpy()

cryptos = ['Ethereum', 'Solana', 'Tether']



for i, crypto in enumerate(cryptos):
    actual_all = test_actual[:, :, i].flatten()
    pred_all = test_pred[:, :, i].flatten()
    
    mae = mean_absolute_error(actual_all, pred_all)
    mse = mean_squared_error(actual_all, pred_all)
    rmse = np.sqrt(mse)
    r2 = r2_score(actual_all, pred_all)
    
    print(f"{crypto}:")
    print(f"MAE:{mae:.4f}")
    print(f"RMSE:{rmse:.4f}")
    print(f"R²:{r2:.4f}")
    print()
```

```python
sample_idx = 0

fig, axes = plt.subplots(3, 1, figsize=(14, 10))

for i, crypto in enumerate(cryptos):
    actual_sample = test_actual[sample_idx, :, i]
    pred_sample = test_pred[sample_idx, :, i]
    
    axes[i].plot(range(1, horizon+1), actual_sample, 'bo-', label='Actual', linewidth=2, markersize=8)
    axes[i].plot(range(1, horizon+1), pred_sample, 'rx-', label='Predicted', linewidth=2, markersize=8)
    axes[i].set_title(f'{crypto} - 10 Day Forecast', fontsize=12)
    axes[i].set_xlabel('Days')
    axes[i].set_ylabel('Price (scaled)')
    axes[i].legend()
    axes[i].grid(True, alpha=0.3)

plt.suptitle('LSTM Predictions vs Actual Values', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
def forecast_future(model, last_sequence, n_days=30, seq_len=60, horizon=10):
    model.eval()
    current_seq = last_sequence.copy()
    forecasts = []
    
    with torch.no_grad():
        for _ in range(n_days // horizon + 1):
            input_tensor = torch.FloatTensor(current_seq).unsqueeze(0)
            pred = model(input_tensor).numpy()[0]
            forecasts.append(pred)
            current_seq = np.vstack([current_seq[horizon:], pred])
    
    return np.vstack(forecasts)[:n_days]


last_sequence = X[-1]  # (60, 3)


forecast_30 = forecast_future(model, last_sequence, n_days=30)


forecast_30_original = scaler.inverse_transform(forecast_30.reshape(-1, 3)).reshape(-1, 3)
```
