```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as seaborn
from collections import Counter
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
# Download latest version
df = pd.read_csv('movies_metadata.csv')
```

```python
df_nuclear = df.copy()
cols_before = df_nuclear.shape[1]


keywords = [
    'homepage', 'status', 'overview', 'poster', 'original', 
    'title', 'belongs', 'collection', 'imdb', 'spoken', 
    'language', 'tagline', 'video', 'popularity'
]

cols_to_remove = []
for col in df_nuclear.columns:
    col_lower = str(col).lower()
    if any(keyword in col_lower for keyword in keywords):
        cols_to_remove.append(col)


df_nuclear = df_nuclear.drop(columns=cols_to_remove)

print(f"Удалено столбцов: {len(cols_to_remove)}")
print(f"Было: {cols_before}, стало: {df_nuclear.shape[1]}")
print(f"Удаленные столбцы:")
for col in cols_to_remove:
    print(f"  '{col}'")
    
print(f"Оставшиеся столбцы ({df_nuclear.shape[1]} шт.):")
for col in df_nuclear.columns:
    print(f"  '{col}'")
```

```python
from sklearn.impute import SimpleImputer
numeric_cols = df_nuclear.select_dtypes(include=np.number).columns
imputer = SimpleImputer(strategy='median')
X_numeric = imputer.fit_transform(df_nuclear[numeric_cols])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_numeric)

print(f"Масштабированные данные: {X_scaled.shape}")
print(f"NaN в данных: {np.isnan(X_scaled).sum()}")
```

```python
inertia = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(2, 11), inertia, 'bo-')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Метод локтя')
plt.show()

# Выбираем k (визуально по графику)
optimal_k = 4  # измените на ваш выбор
```

```python
def recommend_by_title(movie_title, n=5):
    """Рекомендации по названию фильма"""
    # Находим индекс фильма
    if 'title' in df.columns:
        matches = df[df['title'].str.contains(movie_title, case=False, na=False)]
        if len(matches) > 0:
            movie_idx = matches.index[0]
            return recommend_movies(movie_idx, n)
    return "Фильм не найден"

# Пример использования
print(recommend_by_title("Toy Story", 3))
```
