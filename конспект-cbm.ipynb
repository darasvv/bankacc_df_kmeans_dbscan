```python
#batching
import numpy as np

# 1. Генерация данных
np.random.seed(42)  # Для воспроизводимости
X = np.random.randn(1000, 10)
y = np.random.randint(0, 2, 1000)

# 2. Параметры
batch_size = 32

print("Компактный вывод батчей:\n")

# 3. Батчинг
def get_batches(X, y, batch_size):
    indices = np.random.permutation(len(X))
    for i in range(0, len(X), batch_size):
        batch_idx = indices[i:i+batch_size]
        yield X[batch_idx], y[batch_idx]

# 4. Вывод
for i, (X_batch, y_batch) in enumerate(get_batches(X, y, batch_size), 1):
    print(f"Батч {i:2d}: размер={len(X_batch):3d}, "
          f"классы 0/1={np.sum(y_batch==0):3d}/{np.sum(y_batch==1):3d}, "
          f"среднее X={np.mean(X_batch):.3f}")
```

```python
#map-reduce
from collections import defaultdict

# 1. Данные для обработки
data = ["привет мир", "мир большой", "привет всем"]

print("Исходные данные:")
for i, text in enumerate(data, 1):
    print(f"  Строка {i}: '{text}'")

# 2. Mapper функция - преобразует строку в пары (слово, 1)
def mapper(text):
    return [(word, 1) for word in text.split()]

# 3. Reducer функция - суммирует значения для каждого ключа
def reducer(key, values):
    return sum(values)

# 4. MAP фаза
print("\nФаза MAP (преобразование):")
mapped = []
for item in data:
    result = mapper(item)
    mapped.extend(result)
    print(f"  '{item}' -> {result}")

# 5. SHUFFLE фаза (группировка по ключам)
print("Фаза SHUFFLE (группировка):")
shuffled = defaultdict(list)
for key, value in mapped:
    shuffled[key].append(value)
    print(f"  Ключ '{key}' получает значение {value}")

# 6. REDUCE фаза
print("Фаза REDUCE (агрегация):")
result = {}
for key, values in shuffled.items():
    reduced_value = reducer(key, values)
    result[key] = reduced_value
    print(f"  Ключ '{key}' значения {values} -> {reduced_value}")

# 7. Выводим итог
print("Результат:")
for word, count in result.items():
    print(f"  Слово '{word}' встречается {count} раз(а)")
print("="*50)

# Итоговая статистика
total_words = sum(len(text.split()) for text in data)
print(f"\nСтатистика:")
print(f"  Всего слов в текстах: {total_words}")
print(f"  Уникальных слов: {len(result)}")
```
